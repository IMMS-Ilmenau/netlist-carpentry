```python
import os
from netlist_carpentry import read, Circuit

empty = Circuit(name="vhdl_circuit")
c = read("run1_dig/src/META_X_dig_core.vhd", source_paths=["~/oss_cad_suite/environment"])
```

```python
def read_source(path: str):
    if path.endswith(".v"):
        empty.add_from_circuit(path)
    else:
        c = read(path, source_paths=["~/oss_cad_suite/environment"])
        empty.add_from_circuit(c)
```

```python
from pathlib import Path

files = [p for p in Path("run1_dig").rglob("*") if p.is_file()]
print(files)
with open("run1_dig/sources.txt") as f:
    paths = ["run1_dig/"+ln for ln in f.read().splitlines() if ln]
print(paths)
for p in paths:
    read_source(p)
```
